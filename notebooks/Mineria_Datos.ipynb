# ***LLAMADO DE DATASET***

```python
import os
import pandas as pd
from google.colab import drive

# 1. Montar Google Drive
drive.mount('/content/drive')

# 2. Rutas de los archivos
ruta_armenia = '/content/drive/MyDrive/RNT_Armenia_Quindio.csv'
ruta_risaralda = '/content/drive/MyDrive/RNT_Risaralda.csv'


# =========================================================================
# BLOQUE 1: PROCESAMIENTO INDEPENDIENTE - ARMENIA / QUINDÍO
# =========================================================================
print("="*60)
print(" 📍 PROCESANDO ARCHIVO: ARMENIA / QUINDÍO")
print("="*60)

if not os.path.exists(ruta_armenia):
    print(f"❌ Archivo no encontrado en: {ruta_armenia}")
else:
    try:
        df_armenia = pd.read_csv(ruta_armenia, encoding='latin1', sep=';')
        if len(df_armenia.columns) == 1:
            df_armenia = pd.read_csv(ruta_armenia, encoding='latin1', sep='\t')
            print("✓ Cargado con éxito (Separador: Tabulación '\\t')")
        else:
            print("✓ Cargado con éxito (Separador: Punto y coma ';')")
    except Exception as e:
        df_armenia = pd.read_csv(ruta_armenia, encoding='latin1', sep=None, engine='python', on_bad_lines='skip')
        print("✓ Cargado usando detección automática de separador.")

    # Limpiar espacios en los encabezados
    df_armenia.columns = df_armenia.columns.str.strip()

    print(f"Dimensiones: {df_armenia.shape[0]} filas | {df_armenia.shape[1]} columnas")

    print("\n--- Tipos de datos (dtypes) Armenia ---")
    print(df_armenia.dtypes)

    print("\n--- Muestra de datos Armenia ---")
    display(df_armenia.head(3))


# =========================================================================
# BLOQUE 2: PROCESAMIENTO INDEPENDIENTE - RISARALDA (Con codificación utf-8-sig)
# =========================================================================
print("\n" + "="*60)
print(" 📍 PROCESANDO ARCHIVO: RISARALDA")
print("="*60)

if not os.path.exists(ruta_risaralda):
    print(f"❌ Archivo no encontrado en: {ruta_risaralda}")
else:
    try:
        # Usamos utf-8-sig para limpiar el BOM (ï»¿) y corregir la Ñ
        df_risaralda = pd.read_csv(ruta_risaralda, encoding='utf-8-sig', sep=';')
        if len(df_risaralda.columns) == 1:
            df_risaralda = pd.read_csv(ruta_risaralda, encoding='utf-8-sig', sep='\t')
            print("✓ Cargado con éxito (Separador: Tabulación '\\t' | Encoding: utf-8-sig)")
        else:
            print("✓ Cargado con éxito (Separador: Punto y coma ';' | Encoding: utf-8-sig)")
    except Exception as e:
        df_risaralda = pd.read_csv(ruta_risaralda, encoding='utf-8-sig', sep=None, engine='python', on_bad_lines='skip')
        print("✓ Cargado usando detección automática de separador.")

    # Limpiar espacios en blanco en las columnas
    df_risaralda.columns = df_risaralda.columns.str.strip()

    print(f"Dimensiones: {df_risaralda.shape[0]} filas | {df_risaralda.shape[1]} columnas")

    print("\n--- Tipos de datos (dtypes) Risaralda ---")
    print(df_risaralda.dtypes)

    print("\n--- Muestra de datos Risaralda ---")
    display(df_risaralda.head(3))
```

# ***cruce entre las fuentes***

```python
print('--- Columnas de df_armenia antes de la unificación ---')
print(df_armenia.columns.tolist())

print('\n--- Columnas de df_risaralda antes de la unificación ---')
print(df_risaralda.columns.tolist())

# Renombrar columnas de df_risaralda para que coincidan con df_armenia
column_mapping = {
    'CODIGO_MUNICIPIO': 'COD_MUN',
    'CODIGO_DEPARTAMENTO': 'COD_DPTO',
    'NUMERO_DE_HABITACIONES': 'HABITACIONES',
    'NUMERO_DE_CAMAS': 'CAMAS',
    'NUMERO_DE_EMPLEADOS': 'NUM_EMP'
}

df_risaralda_standardized = df_risaralda.rename(columns=column_mapping)

print('\n--- Columnas de df_risaralda después de estandarización ---')
print(df_risaralda_standardized.columns.tolist())

# Verificar si todas las columnas de df_armenia están presentes en df_risaralda_standardized
missing_in_risaralda = set(df_armenia.columns) - set(df_risaralda_standardized.columns)
if missing_in_risaralda:
    print(f"Advertencia: Columnas faltantes en df_risaralda_standardized: {missing_in_risaralda}")
    # Asegurar que df_risaralda_standardized tenga todas las columnas de df_armenia, rellenando con NaN si es necesario
    df_risaralda_standardized = df_risaralda_standardized.reindex(columns=df_armenia.columns)

# Concatenar los DataFrames
initial_rows_armenia = len(df_armenia)
initial_rows_risaralda = len(df_risaralda_standardized)

df_total = pd.concat([df_armenia, df_risaralda_standardized], ignore_index=True)

final_rows_total = len(df_total)

print(f"\nNúmero de filas en df_armenia antes de concatenar: {initial_rows_armenia}")
print(f"Número de filas en df_risaralda antes de concatenar: {initial_rows_risaralda}")
print(f"Número de filas en df_total después de concatenar: {final_rows_total}")

# Mostrar el DataFrame unificado completo
print('\n--- DataFrame unificado completo (df_total) ---')
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
display(df_total)
pd.reset_option('display.max_rows')
pd.reset_option('display.max_columns')
```
